```python
import pandas as pd

weather_data = pd.read_csv('data/plant_weather_data.csv')
weather_data["date_time"] = pd.to_datetime(weather_data["date_time"])
```

```python
group = 3

plant_data = pd.read_csv(
    f"data/plant_generation_data_groupe_{group}.csv",
)
plant_data["date_time"] = pd.to_datetime(plant_data["date_time"])
```

```python
plant_weather_data = pd.merge(
    plant_data[["date_time"]],
    weather_data,
    how="inner",
    left_on=["date_time"],
    right_on=["date_time"],
)

assert len(plant_weather_data) == len(plant_data)
```

```python
plant_weather_data.to_csv(
    f"data/plant_weather_data_groupe_{group}.csv",
    index=False,
)
```

```python
target_data = pd.read_csv(
    f"data/target_groupe_{group}.csv",
)
target_data["date_time"] = pd.to_datetime(target_data["date_time"])
```

```python
plant_weather_forecast_data = pd.merge(
    target_data[["date_time"]],
    weather_data,
    how="inner",
    left_on=["date_time"],
    right_on=["date_time"],
)
```

```python
plant_weather_forecast_data.to_csv(
    f"data/plant_weather_forecast_groupe_{group}.csv",
    index=False,
)
```
