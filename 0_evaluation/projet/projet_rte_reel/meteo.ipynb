```python
import pandas as pd
```

```python
df_1 = pd.read_csv("data/synop.201901.csv", sep=";")
df_2 = pd.read_csv("data/synop.201902.csv", sep=";")
df_3 = pd.read_csv("data/synop.201903.csv", sep=";")
```

```python
mapping_dict = {
    "numer_sta": "station_id",
    "date": "date",
    "pmer": "sea_level_pressure_pa",
    "tend": "pressure_change_3h_pa",
    "cod_tend": "barometric_trend_type_code",
    "dd": "wind_direction_10min_deg",
    "ff": "wind_speed_10min_ms",
    "t": "temperature_k",
    "td": "dew_point_k",
    "u": "humidity_percent",
    "vv": "horizontal_visibility_m",
    "ww": "current_weather_code",
    "w1": "past_weather_1_code",
    "w2": "past_weather_2_code",
    "n": "total_cloud_cover_percent",
    "nbas": "low_cloud_cover_octa",
    "hbas": "low_cloud_base_height_m",
    "cl": "low_cloud_type_code",
    "cm": "mid_cloud_type_code",
    "ch": "high_cloud_type_code",
    "pres": "station_pressure_pa",
    "niv_bar": "barometric_level_pa",
    "geop": "geopotential_m2s2",
    "tend24": "pressure_change_24h_pa",
    "tnN": "min_temperature_n_hours_k",
    "txN": "max_temperature_n_hours_k",
    "tminsol": "min_ground_temperature_12h_k",
    "sw": "measurement_method_code",
    "tw": "wet_bulb_temperature_k",
    "raf10": "wind_gusts_10min_ms",
    "rafper": "wind_gusts_period_ms",
    "per": "gust_measurement_period_min",
    "etat_sol": "ground_state_code",
    "ht_neige": "total_snow_ice_height_m",
    "ssfrai": "fresh_snow_height_m",
    "perssfrai": "fresh_snow_measurement_period_tenths_of_hour",
    "rrN": "precipitation_n_hours_mm",
    "phenspeN": "special_phenomenon_code",
    "nnuageN": "cloud_cover_layer_n_octa",
    "etypeN": "cloud_type_n_code",
    "hnuageN": "cloud_base_height_n_m",
}
```

```python
df = df.rename(mapping_dict, axis=1)
```

```python
df_select = df[
    [
        "station_id",
        "date",
        "total_cloud_cover_percent",
        "low_cloud_cover_octa",
        "temperature_k",
        "humidity_percent",
        "horizontal_visibility_m",
        "wet_bulb_temperature_k",
        "low_cloud_type_code",
        "mid_cloud_type_code",
        "high_cloud_type_code",
    ]
]
```

```python
df_select
```

```python
id_station_mapping = {
    "67005": "Réunion",
    "71805": "Martinique",
    "78890": "Guadeloupe",
    "78922": "Martinique",
    "78925": "Martinique",
    "81401": "Guyane",
    "81405": "Guyane",
    "81408": "Guyane",
    "81415": "Guyane",
}

df_select["station_id"] = df_select["station_id"].astype(str)

df_select["station_name"] = df_select["station_id"].map(id_station_mapping)
```

```python
df_select = df_select.dropna(subset=["station_name"])

# Replace all "mq" values with None in all the dataframe
df_select = df_select.replace("mq", None)
df_select["total_cloud_cover_percent"] = df_select["total_cloud_cover_percent"].astype(
    float
)
df_select["low_cloud_cover_octa"] = df_select["low_cloud_cover_octa"].astype(float)
df_select["temperature_k"] = df_select["temperature_k"].astype(float)
df_select["humidity_percent"] = df_select["humidity_percent"].astype(float)
df_select["horizontal_visibility_m"] = df_select["horizontal_visibility_m"].astype(
    float
)
df_select["wet_bulb_temperature_k"] = df_select["wet_bulb_temperature_k"].astype(float)
```

```python
df_select_gr = df_select.groupby(
    ["station_id", "station_name", "date"], as_index=False
).agg(
    {
        "total_cloud_cover_percent": "mean",
        "low_cloud_cover_octa": "mean",
        "temperature_k": "mean",
        "humidity_percent": "mean",
        "horizontal_visibility_m": "mean",
        "wet_bulb_temperature_k": "mean",
    }
)
```
