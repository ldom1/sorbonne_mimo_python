```python
import pandas as pd
```

```python
df = pd.read_csv(
    "data/production_electricite_par_filiere.csv",
    sep=";",
)
```

```python
df
```

```python
df["Territoire"].unique()

df = df[df["Territoire"].isin(["Guyane", "Réunion", "Martinique", "Guadeloupe"])]
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
import datetime

df_train = df[
    (df["Date"] >= pd.Timestamp(datetime.datetime(2019, 1, 1), tz="UTC"))
    & (df["Date"] < pd.Timestamp(datetime.datetime(2019, 12, 29), tz="UTC"))
]

df_test = df[
    (df["Date"] >= pd.Timestamp(datetime.datetime(2019, 12, 29), tz="UTC"))
    & (df["Date"] < pd.Timestamp(datetime.datetime(2020, 1, 1), tz="UTC"))
]
```

```python
df_train
```

```python
df_test
```

```python
df_train.to_csv(
    "train_production_electricite_par_filiere.csv",
    sep=";",
    index=False,
    header=True,
)
```

```python
df_test.to_csv(
    "test_production_electricite_par_filiere.csv",
    sep=";",
    index=False,
    header=True,
)
```
