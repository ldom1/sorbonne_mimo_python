# Chapitre 1 - Les Bases du Langage

Le langage Python est un langage de programmation polyvalent et largement utilisé, connu pour sa simplicité et sa lisibilité. Il est souvent recommandé pour les débutants en raison de sa syntaxe claire et intuitive, mais il est également puissant et utilisé dans des domaines variés tels que le développement web (exposition de serveurs APIs par exemple), l'analyse de données, l'intelligence artificielle, etc...

La base de l'interaction avec un langage de programmation est **la variable**. Une variable agit comme un conteneur où l'on stocke des informations. Elle permet de manipuler des données de manière dynamique et flexible.

En Python, les variables sont particulièrement faciles à utiliser grâce à la **nature dynamique du langage**. Vous n'avez pas besoin de déclarer le type de données d'une variable à l'avance, ce qui permet une grande flexibilité dans la gestion des données. Cependant, cette flexibilité nécessite également une compréhension claire des différents types de données disponibles et de leur utilisation appropriée.

Ce n'est pas pareil dans les langages que vous avez appris dans les autres cours (Java ou C): Dans ces langages, vous devez déclarer le type de la variable avant de l'utiliser. Par exemple, si vous déclarez une variable comme un entier, vous ne pouvez pas y stocker une chaîne de caractères sans la convertir au préalable.

En Python, c'est possible ! Mais il faut donc être prudent, car cela peut entraîner des erreurs si vous essayez d'effectuer des opérations sur des types de données incompatibles!


En Python, il existe plusieurs types de données:

- La constante None : elle signifie "je n'ai pas de valeur, mon conteneur est vide". C'est l'équivalent de NULL dans d'autres langages.

- Les booléens : True et False.
- Les nombres : en Python, ils peuvent être des entiers, des réels, des complexes, des fractions, etc.
- Les chaînes de caractères : ce sont des messages textuels.
- Les tableaux indicés : les listes et les tuples. Un tableau indicé est comme un casier où chaque case est numérotée de manière continue, **en commençant par 0**.
- Les tableaux associatifs : appelés dictionnaires, ils stockent des paires clé-valeur. Contrairement aux tableaux indicés, les cases sont identifiées par des étiquettes textuelles comme "solar_pv", "wind_turbine", etc. Ces étiquettes sont appelées clés, et les valeurs peuvent être de n'importe quel type de données.

Il existe aussi d'autres types de données, plus avancés:
- Les ensembles : ce sont des collections non ordonnées d'éléments uniques: set()
- Les fichiers : ce sont des objets qui permettent de lire et d'écrire des données sur le disque dur.

## Les chaînes de caractères

Une chaîne de caractères représente un message textuel. On appelle une chaine de caractères une *string* en anglais.
En Python, les chaînes de caractères sont des séquences de caractères entourées de guillemets. Elles peuvent contenir des lettres, des chiffres, des espaces et d'autres symboles.

On peut les délimiter de 4 manières:

* Avec des guillemets : `"`
* Avec des apostrophes : `'`
* Avec des triples guillemets : `"""`
* Avec des triples apostrophes : `'''`

Les triples guillemets permettent d'écrire des chaînes de caractères sur plusieurs lignes, ce qui est très pratique pour les longs messages ou les documents.

On peut aussi utiliser des guillemets simples à l'intérieur de guillemets doubles et vice versa.

Les opérations de base sur les chaînes de caractères incluent la concaténation, l'indexation, le slicing... 

### Les opérations sur les chaînes de caractères

````python
# Création de la variable var
a = "Descriptif : « C’est un roc ! … c’est un pic ! … c’est un cap !"
````

````python
print(a)
````

Nous venons de créér une variable `a` contenant une chaîne de caractères. Nous avons ensuite utilisé la fonction `print` pour afficher le contenu de cette variable à l'écran.

Il est possible de créer cette chaîne de caractères de 4 manières différentes, en utilisant des guillemets simples, des guillemets doubles, des triples guillemets ou des triples apostrophes.

La façon la plus standard est d'utiliser des guillemets doubles.

````python
a1 = "Descriptif : « C’est un roc ! … c’est un pic ! … c’est un cap !"
a2 = "Descriptif : « C’est un roc ! … c’est un pic ! … c’est un cap !"
a3 = """Descriptif : « C’est un roc ! … c’est un pic ! … c’est un cap !"""
a4 = """Descriptif : « C’est un roc ! … c’est un pic ! … c’est un cap !"""

print("a1 == a2 ? ", a1 == a2, "a1 et a2 sont de types:", type(a1), type(a2))
print("a1 == a3 ? ", a1 == a3, "a1 et a3 sont de types:", type(a1), type(a3))
print("a1 == a4 ? ", a1 == a4, "a1 et a4 sont de types:", type(a1), type(a4))
print("a2 == a3 ? ", a2 == a3, "a2 et a3 sont de types:", type(a2), type(a3))
````

### Formatage d'une chaîne

En Python il est possible de formater une chaîne pour y insérer la valeur de variables de plusieurs manières:
* L'opérateur `%` semblable au `printf` du langage C  
* La méthode `"chaine".format()`  
* Les f-strings avec le préfixe `f`  

##### L'opérateur `%`  

Syntaxe proche de `printf` en C :  
`"texte %<format1>, autre texte %<format2>" % (valeur1, valeur2)`  

Exemples de formats :  
- `%s` : texte  
- `%d` : entier  
- `%f` : réel  

##### La méthode `.format()`  

Permet d'insérer des variables dans une chaîne :  
`"texte {} autre texte {}".format(valeur1, valeur2)`  

##### Les f-strings  

Introduites en Python 3.6, elles permettent d'insérer directement des variables :  
`f"texte {variable}"`


````python
# Avec l'opérateur %
d1 = (
    "Curieux : « De quoi sert cette oblongue capsule ? D’écritoire, monsieur, ou de boîte à %s ? »"
    % "couteaux"
)
d2 = (
    "Curieux : « De quoi sert cette oblongue capsule ? D’écritoire, monsieur, ou de boîte à %s ? »"
    % "ciseaux"
)
d3 = (
    "Curieux : « De quoi sert cette oblongue capsule ? D’écritoire, monsieur, ou de boîte à %s ? »"
    % 1000
)
print(d1)
print(d2)
print(d3)
````

````python
# Avec la méthode format()
d1 = "Curieux : « De quoi sert cette oblongue capsule ? D’écritoire, monsieur, ou de boîte à {0}, {1}, {2} ? »".format(
    "fourchette", "cuillères", 2000
)
print(d1)
````

````python
# Avec la méthode f-string : Python 3.6 ou plus
elt1 = "baguettes"
elt2 = "ciseaux"
elt3 = 5000

d1 = f"Curieux : « De quoi sert cette oblongue capsule ? D’écritoire, monsieur, ou de boîte à {elt1}, {elt2}, {elt3} ? »"
print(d1)
````

La fonction `format` propose une autre manière de formatter.

````python
a, b = 10, 20
var = "a={0}, b={1}"
print(var.format(a, b))
print("a={0}, b={1}".format(a, b))
print("a={0:04d}, b={1:08.2f}".format(a, b))
print("a={qte:04d}, b={prix:08.2f}".format(prix=b, qte=a))
````

Depuis la version de Python 3.6 on peut aussi préfixer une chaîne par un `f`, dans ce cas les expressions entre accolades qu'elle contient seront remplacées par leur valeur.

Enfin depuis Python 3.8 la syntaxe a été entendue: ***COMMENTAIRE : A été entendue ? Je ne comprends pas cette phrase...***

````python
print("{a=}, {b=}")
print(f"{a=}, {b=}")
print(f"{a=:04d}, {b=:08.2f} {a+b*1.0=}")
````

## Listes et tuples

En **Python**, il existe deux types principaux de tableaux indicés :

- Listes : Délimitées par des crochets [ ], elles peuvent contenir divers éléments et sont modifiables après leur création.
- Tuples : Délimitées par des parenthèses ( ), elles peuvent également contenir divers éléments mais sont immuables une fois créées.

Les listes sont TRES utiles pour stocker des collections d'objets, car elles peuvent contenir des éléments de types différents. Par exemple, vous pouvez avoir une liste contenant des entiers, des chaînes de caractères et même d'autres listes.

````python
# Liste
ma_liste = [1, 2, 3, 4, 5]
print(ma_liste)
````

````python
# Il est possible de créer des listes de n'importe quel type
ma_liste = [1, "a", True, [1, 2, 3]]
print(ma_liste)
````

On accède aux éléments du tableau par la notation `<nom variable tableau>[<numéro du casier>]`

````python
print(ma_liste[0])
print(ma_liste[1])
print(ma_liste[2])
````

Une case du tableau peut contenir un autre tableau! On enchaîne alors la sélection avec les crochets.

````python
print(ma_liste[3][2])
````

En python on peut utiliser des indices négatifs:

* `-1` sera le dernier élément
* `-2` sera l'avant dernier élément

````python
print(ma_liste[-1])
print(ma_liste[-1][-2])
````

##### Quelques opérations utiles avec les listes

- ajouter un élément à la fin de la liste: append
- inserer un élément à une position donnée: insert
- supprimer un élément à une position donnée: del
- extraire un élément à une position donnée: pop
- trier la liste: sort
- inverser la liste: reverse
- compter le nombre d'éléments: count
- obtenir la longueur de la liste: len
- vérifier si un élément est dans la liste: in
- obtenir l'index d'un élément: index

````python
var = [10, 20, 30, 40, 50, 60]
print("Nombre d'éléments:", len(var))
print(len("MIMO"))
# Ajouter en fin
var.append(70)
print(var)
````

````python
# Insertion au milieu
var.insert(2, 25)
print(var)
````

````python
# Suppression d'un élément
del var[2]  # supprime l'élément à l'indice 3
print(var)
````

````python
# Extraire un élément à une position
v = var.pop(1)  # retire l'élément à l'indice 1
print(v)
print(var)
````

````python
# Trier une liste
var = [20, 30, 10, 50, 40]
print("La liste:", var)
var.sort()  # trie la liste
print("La liste triée:", var)
var.sort(reverse=True)  # trie la liste dans l'autre sens
print("La liste triée dans l'autre sens:", var)
````

````python
# Compter le nombre d'éléments
var = [1, 2, 3, 4, 5, 1, 2, 3]
print("La liste:", var)
print("Nombre d'éléments:", len(var))
print("Nombre d'éléments 1:", var.count(1))
````

````python
# Est-ce qu'un élément est dans la liste ?
print("1 est-il dans la liste ?", 1 in var)  # True
print("10 est-il dans la liste ?", 10 in var)  # False
````

````python
# Obtention de l'indice d'un élément
var = [1, 2, 3, 4, 5]
print("L'indice de 1 est:", var.index(1))  # 0
print("L'indice de 3 est:", var.index(3))  # 2
````

````python
# Concaténation de listes & répétition: Cela fonctionne comme pour les chaînes de caractères
a = [1, 2, 3]
b = [4, 5, 6]
print(a + b)  # Concaténation de deux listes
print(a * 2)  # Répétition de la liste
````

````python
# Tuple
mon_tuple = (1, 2, 3, 4, 5)
print(mon_tuple)

mon_tuple[3] = 2
````

**Question: Mais alors, à quoi sert le tuple ?**

Il n'apporte rien de plus que la liste et l'on pourrait s'en passer. Mais comme il n'est pas modifiable, sa structure de données (les octets utilisés pour le réprésenter en mémoire) est moins gourmande que celle d'une liste et il devient intéressant pour cela.

## Les booléens

Les booléens en Python sont des valeurs qui peuvent être soit True (vrai) soit False (faux). Ils jouent un rôle crucial dans les opérations logiques et les structures conditionnelles. Voici les règles qui les régissent :

Un opérande booléen est une expression qui retourne True ou False. Par exemple, `5 > 3` est un opérande booléen qui retourne True.
Les opérateurs de comparaison (comme `==`, `!=`, `<`, `>`, `<=`, `>=`) retournent également des valeurs booléennes. Par exemple, `5 == 5` retourne True, tandis que `5 != 3` retourne True.

### Opérateurs Booléens

#### 1. Ordre de Priorité :

* not : Cet opérateur a la priorité la plus élevée. Il inverse la valeur booléenne de son opérande.
* and : Cet opérateur vient ensuite. Il retourne True uniquement si toutes les conditions sont vraies.
* or : Cet opérateur a la priorité la plus basse. Il retourne True dès qu'une des conditions est vraie.

#### 2. Évaluation Progressive

* Les opérateurs booléens sont dits "progressifs" ou "court-circuités". Cela signifie qu'ils arrêtent leur évaluation dès que le résultat est déterminé.
* Par exemple, dans une expression **A or B, si A est True, B ne sera pas évalué car le résultat est déjà connu**.

#### 3. Retour des Opérandes :

* Les opérateurs booléens retournent l'une de leurs opérandes, celle qui a permis de déterminer le résultat.
* Par exemple, dans A and B, si A est False, l'expression retourne A.

### Conversion en Booléen
Python peut convertir n'importe quelle valeur en booléen selon les règles suivantes :

* False : Toutes les valeurs "vides" ou "nulles" sont considérées comme False. Cela inclut :
    * None
    * 0 (zéro)
    * Chaînes de caractères vides ("")
    * Listes, tuples, dictionnaires, et autres collections vides

* True : Toutes les autres valeurs sont considérées comme True.

### Exemples d'Utilisation

* or : Retourne True dès qu'une des conditions est vraie.
    * Exemple : True or False retourne True.
* and : Retourne True uniquement si toutes les conditions sont vraies.
    * Exemple : True and False retourne False.
* not : Retourne l'inverse de la valeur booléenne.

* Exemple : not True retourne False.

Ces règles permettent de manipuler efficacement les expressions logiques en Python, en tirant parti de la nature court-circuitée des opérateurs pour optimiser les calculs.

````python
# Initialisation des variables
a = 0  # False
b = "Bonjour"  # True
c = [1, 2, 3]  # True

# Évaluation avec 'and'
resultat_and = a and b and c
print("Résultat avec 'and' :", resultat_and)

# Évaluation avec 'or'
resultat_or = a or b or c
print("Résultat avec 'or' :", resultat_or)

# Inversion avec 'not'
resultat_not = not a
print("Résultat avec 'not' :", resultat_not)
````

Les caractéristiques des ensembles sont les suivantes :
- Ils ne contiennent pas de doublons.
- Les integers & floats sont considérés comme des objets immuables.

Les ensembles sont utiles pour effectuer des opérations mathématiques telles que l'union, l'intersection et la différence entre deux ensembles. Ils sont également utilisés pour vérifier l'appartenance d'un élément à un ensemble.

````python
set_1 = {1, 2, 3}
set_2 = {3, 4, 5}
print(set_1 | set_2)  # Union
print(set_1 & set_2)  # Intersection
print(set_1 ^ set_2)  # Différence symétrique
````

## Les fichiers
Les fichiers sont des objets qui permettent de lire et d'écrire des données sur le disque dur. Ils sont utilisés pour stocker des informations de manière permanente. En Python, vous pouvez ouvrir un fichier en utilisant la fonction `open()`, puis lire ou écrire des données à l'aide de méthodes telles que `read()`, `write()`, et `close()`.

````python
# Ecrire un fichier texte
file_path = "exemple.txt"
with open(file_path, "w") as file:
    file.write("Bonjour, ceci est un exemple de fichier texte.\n")
    file.write("Il contient plusieurs lignes de texte.\n")
    file.write("C'est un exemple simple.\n")
````

````python
# Lire le fichier texte
with open(file_path, "r") as file:
    contenu = file.read()
    print("Contenu du fichier :")
    print(contenu)
````
