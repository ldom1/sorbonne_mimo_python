## Chaînes de caractères — Corrigé

### Exercice 🐍 — Chaînes de caractères

##### Objectifs
- Créer des chaînes, concaténation, slicing, f-strings

##### Instructions
1. Créez une variable `prenom` et une variable `nom`. Formatez une chaîne pour afficher "Bonjour, **prenom nom**." avec une f-string.
2. Créez `phrase = "Python est génial"`. Affichez les 6 premiers caractères, puis les 6 derniers (en utilisant le slicing).
3. Concaténez `prenom` et `nom` avec un espace entre les deux et affichez la longueur de la chaîne résultante.


```python
prenom, nom = "Alice", "Dupont"
print(f"Bonjour, {prenom} {nom}.")

phrase = "Python est génial"
print(phrase[:6], phrase[-6:])

complet = prenom + " " + nom
print(len(complet))
```

## Listes — Corrigé

### Exercice 🐍 — Listes

##### Objectifs
- Créer une liste, accéder aux éléments (indices, slicing)
- Utiliser `append`, `len`, `in` ou `index`

##### Instructions
1. Créez une liste `fruits` contenant au moins 4 noms de fruits.
2. Affichez le premier et le dernier élément (en utilisant un indice négatif pour le dernier).
3. Affichez une sous-liste du 2e au 4e élément (slicing).
4. Ajoutez un fruit avec `append`. Affichez True si `"pomme"` est dans la liste, False sinon.


```python
fruits = ["pomme", "banane", "orange", "kiwi"]
print(fruits[0], fruits[-1])
print(fruits[1:4])
fruits.append("mangue")
print("pomme" in fruits)
```

##### Instructions

### Exercice 🐍 — Booléens

##### Objectifs

##### Instructions
