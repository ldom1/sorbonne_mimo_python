```python
##### I - Calculer la Population Totale

my_dep = "76"

# Calculer la population totale du département
pop_dep = df[df.dep_code == my_dep].population.sum()
print("La population du département", my_dep, "est de", pop_dep)

# Calculer la population totale des départements de Haute-Normandie: 27, 76
pop_dep = df[df.dep_code.isin(["27", "76"])].population.sum()

print("La population des départements 27 et 76 est de", pop_dep)
```

```python
##### II - Altitude Maximale

# Calculer l'altitude maximale
max_alt = df.altitude_maximale.max()
print("L'altitude maximale est de", max_alt, "m")

# Lister les villes à cette altitude
df[df.altitude_maximale >= max_alt]
```

```python
# III - Villes les Plus Peuplées

# Lister les 10 villes les plus peuplées
df.sort_values("population", ascending=False).head(10)
```

```python
# Alternative: Utiliser nlargest pour lister les 10 villes les plus peuplées
df.nlargest(10, "population")
```

### Exercice de cartographie
La librairie folium permet d'afficher des fonds de cartes à la GoogleMap/OpenStreetMap

La librairie Nominatim permet quant-à elle de localiser votre position de multiples manières (adresse postale, ip, ...)

1. Créer un jeu de données contenant uniquement les colonnes: nom_standard, latitude_mairie, longitude_mairie, population, grille_densite_texte

2. Centrer la carte sur la moyenne des coordonnées des villes de France

3. Utiliser les colonnes latitude_mairie et longitude_mairie

4. Lister 5 villes au hasard parmis les 10 plus peuplées et 5 villes au hasard parmis les 10 moins peuplées

5. Puis à l'aide de la fonction df.itertuples() afficher un marker par ville avec l'information de la densité de population

```python
# 1. Créer un jeu de données contenant uniquement les colonnes nécessaires
df_selected = df[
    [
        "nom_standard",
        "latitude_mairie",
        "longitude_mairie",
        "population",
        "grille_densite_texte",
    ]
]
```

```python
# Afficher les premières lignes du jeu de données
df_selected.head()
```

```python
# 2. Centrer la carte sur la moyenne des coordonnées des villes de France
#
# Calculer la moyenne des coordonnées des villes de France
import folium


lat_mean, long_mean = (
    df_selected.latitude_mairie.mean(),
    df_selected.longitude_mairie.mean(),
)

# Créer une carte centrée sur la moyenne des coordonnées
m = folium.Map(location=(lat_mean, long_mean), zoom_start=6)
url = "https://python-visualization.github.io/folium/_images/folium_logo.jpg"

# Sélectionner 5 villes au hasard parmis les 10 plus peuplées
top10 = df_selected.nlargest(10, "population")
random5top10 = top10.sample(5)

# Sélectionner 5 villes au hasard
random5 = df_selected.sample(5)  # 5 villes aléatoires

# Ajouter des markers pour les villes sélectionnées
for row in random5top10.itertuples():
    pos = folium.Marker(
        location=(row.latitude_mairie, row.longitude_mairie),
        tooltip=f"{row.nom_standard}: {row.grille_densite_texte}",
        popup=f"<b>Population</b>:{row.population}",
        icon=folium.Icon(color="red"),
    )
    pos.add_to(m)

for row in random5.itertuples():
    pos = folium.Marker(
        location=(row.latitude_mairie, row.longitude_mairie),
        tooltip=f"{row.nom_standard}: {row.grille_densite_texte}",
        popup=f"<b>Population</b>:{row.population}",
        icon=folium.Icon(color="green"),
    )
    pos.add_to(m)

m
```
