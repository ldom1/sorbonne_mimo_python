### Contexte

Jean-Marc Jancovici met en avant les tonnes de marchandises transportées comme un indicateur macroéconomique pertinent pour évaluer la situation économique physique, en complément ou en remplacement du PIB, qu’il considère de moins en moins représentatif des flux physiques. Voici un résumé des ressources et idées principales :

##### Les Tonnes Transportées comme Indicateur
- **Indicateur physique** : Jancovici propose d’utiliser des indicateurs concrets comme les tonnes de marchandises transportées (notamment par camions) pour mieux refléter l’état réel de l’économie. Une baisse de ces tonnes indique une diminution des biens produits et consommés, ce qui reflète une forme de récession physique en Europe depuis plusieurs années.

- **Lien avec la décroissance** : Il associe cette baisse au stress énergétique et à la désindustrialisation, qui sont des signes d’une décroissance s’accélérant avec les crises énergétiques et géopolitiques.

#####  Contexte Économique et Énergétique
- **Relation énergie-économie** : Selon Jancovici, il existe une corrélation quasi linéaire entre la consommation énergétique et la croissance économique. La diminution des flux physiques (comme les tonnes transportées) est donc liée à une baisse de l’énergie disponible, notamment fossile.

- **Critique du PIB** : Le PIB ne prend pas en compte les flux physiques ni les limites planétaires. Jancovici plaide pour des indicateurs intégrant ces dimensions, car ils permettraient de mieux comprendre les impacts économiques réels du changement climatique et de l’épuisement des ressources.

#####  Données sur le Transport de Marchandises
Les statistiques montrent des tendances variées dans le transport terrestre (routier, ferroviaire, fluvial). Par exemple :

- En 2018, le transport routier représentait 89 % du fret terrestre en France et augmentait légèrement (+2,9 %), tandis que le ferroviaire chutait (-4,2 %).
- Ces données illustrent l’importance du transport routier dans l’économie physique actuelle mais aussi sa dépendance aux énergies fossiles.

#### Affirmation

D'après Jean-Marc Jancovici, on observerait une baisse des tonnes de marchandises transportées en Europe depuis plusieurs années, ce qui serait un signe de récession physique et de décroissance économique. Cette tendance serait liée à des facteurs énergétiques et géopolitiques.

#### Objectif

L'objectif est de vérifier cette affirmation en utilisant des données sur le transport de marchandises en France et en Europe pour évaluer l'évolution des tonnes transportées et leur corrélation avec d'autres indicateurs économiques.

### Qu'est-ce que le jeu de données contient?

Le jeu de données contient des informations sur les tonnes-kilomètres de marchandises transportées par région d'arrivée en France. Il est structuré en 6 colonnes :

- **tonnes_kilometres_de_marchandises_transportees_par_region_d_arrivee_transport_marchandise_destination_region**: Quantité totale de marchandises transportées, mesurée en tonnes-kilomètres, pour une région d'arrivée spécifique. Elle prend en compte le trafic intérieur (sur le territoire français) ainsi que le trafic international entrant et sortant. Le trafic en transit et le trafic non commercial ne sont pas inclus

- **date_mesure.year**: Année de la mesure des données. Elle permet de situer temporellement les informations sur le transport de marchandises.

- **date_mesure**: Date exacte à laquelle les données de transport ont été mesurées. Elle peut inclure le jour, le mois et l'année, offrant ainsi une granularité temporelle précise.

- **geocode_region**: Code géographique de la région, conformément à la Nomenclature des unités territoriales statistiques (NUTS). Ce code permet d'identifier de manière unique chaque région, facilitant ainsi l'analyse géographique des données.

- **mode_transport**: Mode de transport utilisé pour acheminer les marchandises (par exemple, routier, ferroviaire, maritime, etc.). Elle aide à comprendre la répartition des marchandises transportées selon les différents modes.

- **libelle_region**: Libellé de la région d'arrivée des marchandises. Elle permet une interprétation plus intuitive des données géographiques par rapport aux codes NUTS.

```python
# Renommons les colonnes pour plus de clarté

# Questions : En utilisant un dictionnaire, renommer les colonnes de la manière suivante :
# - "tonnes_kilometres_de_marchandises_transportees_par_region_d_arrivee_transport_marchandise_destination_region" en "tonnes_km"
# - date_mesure.year en "annee"
# - date_mesure en "date"
# Laisse les autres colonnes inchangées

columns_names_mapping = {
    "tonnes_kilometres_de_marchandises_transportees_par_region_d_arrivee_transport_marchandise_destination_region": "tonnes_km",
    "date_mesure.year": "annee",
    "date_mesure": "date",
}

df = df.rename(columns=columns_names_mapping)
df.head()
```

```python
# On observe que la colonne libelle_region contient à la fois des régions françaises et des pays étrangers. On va donc garder uniquement les régions françaises métropolitaines.

list_regions_francaises = [
    "Auvergne-Rhône-Alpes",
    "Bourgogne-Franche-Comté",
    "Bretagne",
    "Centre-Val de Loire",
    "Corse",
    "Grand Est",
    "Hauts-de-France",
    "Île-de-France",
    "Normandie",
    "Nouvelle-Aquitaine",
    "Occitanie",
    "Pays de la Loire",
    "Provence-Alpes-Côte d'Azur",
]

print(
    f"Nous prenons en compte les {len(list_regions_francaises)} régions françaises métropolitaines."
)
```

```python
# Nous allons donc filtrer le dataframe pour ne garder que les lignes où libelle_region est dans list_regions_francaises

df_fr_metro = df[df["libelle_region"].isin(list_regions_francaises)]
df_fr_metro.head()
```

### 3. Analyse temporelle des données

```python
# Convertis la date en datetime

df_fr_metro["date"] = pd.to_datetime(df_fr_metro["date"])

# Question: Affiche la date minimale et maximale
print("Date minimale:", df_fr_metro["date"].min())
print("Date maximale:", df_fr_metro["date"].max())
```

```python
# Ajoutons une colonne "annee", qui contient l'année de la date
df_fr_metro["annee"] = df_fr_metro["date"].dt.year
```

```python
df_fr_metro
```

```python
# Calculer le nombre total de tonnes-km par année en France et par mode_transport: convertir les tonnes_km en milliard de tonnes-km dans une nouvelle colonne "tonnes_km_milliard"

df_fr_metro["tonnes_km_milliard"] = df_fr_metro["tonnes_km"] / 1_000_000_000

# Affiche le nombre total de tonnes-km par année en France
df_fr_metro.groupby(["annee"], as_index=False).agg(
    {"tonnes_km": "sum", "tonnes_km_milliard": "sum"}
)
```

```python
# On observe que les années 2011, 2012 et 2012 ne sont pas complètes. On va donc les supprimer.
df_fr_metro = df_fr_metro[df_fr_metro["annee"] > 2013]
```

```python
# Affiche le nombre total de tonnes-km par année en France et par mode_transport
df_fr_metro.groupby(["annee", "mode_transport"], as_index=False).agg(
    {"tonnes_km": "sum", "tonnes_km_milliard": "sum"}
)
```

```python
# Et en 2018 ?
df_fr_metro[df_fr_metro["annee"] == 2018].groupby(
    ["annee", "mode_transport"], as_index=False
).agg({"tonnes_km": "sum", "tonnes_km_milliard": "sum"})
```

```python
# Afficher un lineplot du nombre total de tonnes-km par année en France pour le transport routier
import matplotlib.pyplot as plt
df_fr_metro[df_fr_metro["mode_transport"] == "routier"].groupby(
    ["annee"], as_index=False
).agg({"tonnes_km": "sum", "tonnes_km_milliard": "sum"}).plot(
    x="annee", y="tonnes_km_milliard", kind="line"
)
plt.show()
```

```python
# En nous basant sur ce graphique, on peut dire que le transport routier est en augmentation en France.
df_fr_metro_routier = df_fr_metro.groupby(["annee"], as_index=False).agg(
    {"tonnes_km_milliard": "sum"}
)
```

```python
df_fr_metro_routier
```

```python
# Calculons maintenant la tendance du transport routier en France

# Question: Utilise la régression linéaire pour calculer la tendance du transport routier en France
from sklearn.linear_model import LinearRegression

# Nous allons utiliser la régression linéaire pour calculer la tendance du transport routier en France

X = df_fr_metro_routier["annee"].values.reshape(-1, 1)
y = df_fr_metro_routier["tonnes_km_milliard"].values
```

```python
X
```

```python
model = LinearRegression()
```

```python
model
```

```python
model.fit(X, y)
```

```python
model.predict(X)
```

```python
# Nous allons afficher sur un graphique la tendance du transport routier en France avec les données et la droite de régression

plt.figure(figsize=(15, 5))
plt.plot(
    df_fr_metro_routier["annee"],
    df_fr_metro_routier["tonnes_km_milliard"],
    color="blue",
)
plt.plot(df_fr_metro_routier["annee"], model.predict(X), color="red")
plt.xlabel("Année")
plt.ylabel("Tonnes-km (milliard)")
plt.title("Tendance du transport routier en France")
plt.show()
```
