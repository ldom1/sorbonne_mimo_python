````python
# Avec match case
print("Bienvenue dans le Système de Recommandation de Films !")

# Demander les préférences de l'utilisateur
genre = (
    input("Quel genre de film préférez-vous ? (Action, Comédie, Science-fiction) : ")
    .strip()
    .lower()
)
plus_de_2h = (
    input("Souhaitez-vous un film de plus de 2 heures ? (Oui/Non) : ").strip().lower()
)

# Convertir la réponse en booléen
is_plus_de_2h = plus_de_2h == "oui"

# Recommandation en fonction des préférences
match (genre, is_plus_de_2h):
    case ("action", True):
        print("Je vous recommande 'Mad Max: Fury Road'.")
    case ("action", False):
        print("Je vous recommande 'Kill Bill Vol. 1'.")
    case ("comédie", True):
        print("Je vous recommande 'Hot Fuzz'.")
    case ("comédie", False):
        print("Je vous recommande 'Monty Python - Sacré Graal !'.")
    case ("science-fiction", True):
        print("Je vous recommande 'Matrix'.")
    case ("science-fiction", False):
        print("Je vous recommande 'Bienvenue à Gattaca'.")
    case _:
        print("Désolé, je n'ai pas de recommandation pour ces préférences.")
        print(f"Pour: ({genre}, {is_plus_de_2h})")
# Fin du programme
````

## Exercice 🐍 - Jeu de dés

L'objectif de cet exercice est de se familiariser avec les boucles et les instructions conditionnelles. Vous allez créer un petit programme qui simule un jeu de dés.

Le jeu consiste à lancer 2 dés à 6 faces. 2 joueurs s'affrontent tour à tour. Le joueur 1 est paire: s'il obtient 2 nombres paires, il gagne. Le joueur 2 est impair: s'il obtient 2 nombres impairs, il gagne.

Tant qu'aucun joueur n'a gagné, le jeu continue.

### Enoncé

1. Lancer 2 dés à 6 faces en utilisant la fonction `random.randint(1, 6)`.
2. Vérifier si le joueur 1 a gagné (2 nombres paires) ou si le joueur 2 a gagné (2 nombres impairs).
3. Si aucun joueur n'a gagné, recommencer le lancer de dés.
4. Si un joueur a gagné, sortir un message indiquant le gagnant et le nombre de lancers effectués.

````python
import random

joueur_1_gagne = False
joueur_2_gagne = False
count = 0

while not joueur_1_gagne and not joueur_2_gagne:
    # Tour de joueur 1
    de_1 = random.randint(1, 6)
    de_2 = random.randint(1, 6)
    count += 1
    print(f"Joueur 1 a tiré {de_1} et {de_2}")

    if de_1 % 2 == 0 and de_2 % 2 == 0:
        joueur_1_gagne = True
        print("Joueur 1 a gagné !")
        break

    # Tour de joueur 2
    de_1 = random.randint(1, 6)
    de_2 = random.randint(1, 6)
    count += 1
    print(f"Joueur 2 a tiré {de_1} et {de_2}")
    if de_1 % 2 == 1 and de_2 % 2 == 1:
        joueur_2_gagne = True
        print("Joueur 2 a gagné !")

print("Les joueurs ont joué", count, "fois.")
````

````python
## Exercice 🐍 2 - Jeu de Cartes
import random

print("Bienvenue dans le Jeu de Cartes !")
print("Devinez la couleur de la carte tirée !")

liste_de_cartes = [
    "As",
    "2",
    "3",
    "4",
    "5",
    "6",
    "7",
    "8",
    "9",
    "10",
    "Valet",
    "Dame",
    "Roi",
]
paquet = {
    "coeur": liste_de_cartes,
    "carreau": liste_de_cartes,
    "trefle": liste_de_cartes,
    "pique": liste_de_cartes,
}

# Tirer une carte aléatoire
couleur_random = random.choice(list(paquet.keys()))
valeur_random = random.choice(paquet[couleur_random])
carte_random = valeur_random + " de " + couleur_random

# Initialiser la variable de réponse
reponse_correcte = False
nb_tentatives = 0
max_tentatives = 2

while reponse_correcte is False:
    couleur = input(
        "Quelle est la couleur de la carte ? (coeur, carreau, trefle, pique) : "
    )
    couleur = couleur.strip().lower()

    if couleur == couleur_random:
        print(
            "Bravo ! Vous avez deviné la bonne couleur, la carte tirée est :",
            carte_random,
        )
        reponse_correcte = True
    else:
        print(f"Désolé, {couleur} n'est pas la bonne couleur. Essayez encore.")

    nb_tentatives += 1

    if nb_tentatives >= max_tentatives:
        print(
            "Vous avez épuisé toutes vos tentatives. La carte tirée était :",
            carte_random,
        )
        break
# Fin du programme
````

````python
# Exercice 3

alphabet = "abcdefghijklmnopqrstuvwxyz"

message = """
jvttl jvuclub, sl zfzaltl kl zljbypal kl s bupclyzpal whypz whuaolvu-zvyivuul zlyh wpyhal jl sbukp.
uvbz hssvuz uvbz puayvkbpyl khuz sl zfzaltl kl zljbypal cph s vykpuhalby kl sh ipispvaolxbl.
wsbz wlyzvuul ul wvbyyh hjjlkly hb zfzaltl pumvythapxbl la uvbz wvbyyvuz cvsly avbalz slz kvuullz kl s bupclyzpal.

h ipluava,
cshk.
"""

# Déchiffrez le message

# Le mot sorbonne est dans le message !!
indice = 0
message_dechiffre = ""

while "sorbonne" not in message_dechiffre:
    message_dechiffre = ""
    for char in message:
        if char in alphabet:
            index_in_alphabet = alphabet.index(char)
            index_cible = (index_in_alphabet + indice) % len(alphabet)
            message_dechiffre += alphabet[index_cible]
        else:
            message_dechiffre += char
    indice += 1
    if indice > len(alphabet):
        print("Le message n'a pas pu être déchiffré.")
        break

print(f"Indice du chiffrage: {indice}, message déchiffré: {message_dechiffre}")
````

## Exercice 🐍 - L'algorithme d'Euclide - Le calcul du PGCD

L'algorithme d'Euclide est un algorithme efficace pour calculer le plus grand commun diviseur (PGCD) de deux entiers. Le PGCD est le plus grand entier qui divise à la fois les deux nombres sans laisser de reste. Cet algorithme date de 300 avant J.-C. et est toujours utilisé aujourd'hui en mathématiques et en informatique.

2 approches:
- Par différence
- Par division euclidienne

### Par différence

L'algorithme d'Euclide par différence repose sur le principe que le PGCD de deux nombres reste le même si l'on soustrait le plus petit nombre du plus grand. Voici comment cela fonctionne :
1. On commence avec deux entiers positifs a et b.
2. On soustrait le plus petit nombre du plus grand.
3. On répète cette opération jusqu'à ce que les deux nombres soient égaux.
4. Le résultat est le PGCD.

### Par division euclidienne

L'algorithme d'Euclide par division repose sur le principe que le PGCD de deux nombres a et b est le même que le PGCD de b et du reste de la division de a par b. Voici comment cela fonctionne :
1. On commence avec deux entiers positifs a et b.
2. On divise a par b et on obtient le reste r.
3. On remplace a par b et b par r.
4. On répète cette opération jusqu'à ce que b soit égal à 0.
5. Le PGCD est alors le dernier reste non nul.

## Enoncé

Implez l'algorithme d'Euclide par différence et par division euclidienne pour calculer le PGCD de deux entiers positifs a et b.


````python
# Par différence
a = 54
b = 9

# Calculer le PGCD
while b != a:
    if a > b:
        a = a - b
    else:
        b = b - a
    print("Valeurs intermédiaires : a =", a, ", b =", b)

print("Le PGCD est :", b)
````

````python
# Par division
a = 1086
b = 156

# Calculer le PGCD
while b != 0:
    r = a % b
    a = b
    b = r
    print("Valeurs intermédiaires : a =", a, ", b =", b)

print("Le PGCD est :", a)
````
