# Programmation Orientée Objets (POO)

Nous proposons de découvrir les bases de la programmation orientée objets, de manière sobre et légère.

L'idée n'est pas de donner un cours magistral sur les concepts liés à ce type de programmation, mais d'en comprendre la syntaxe et mécanique de base, car **en Python, tout est objet**

![POO](https://www.physik.uzh.ch/~python/python_2015-01/lecture2/slides/images/oopschema.jpg)

## Les classes

Le principe de base de la POO est de permettre de définir de nouveaux types de données que l'on appelle des classes.

Par exemple une personne possède :

* un nom
* un prénom
* un age

Nous pourrions la décrire avec un simple dictionnaire :

````python
personne = {"NOM": "Martin", "PRENOM": "Toto", "AGE": 10}

print(personne, type(personne))
````

Mais une personne peut aussi réaliser des actions :

* respirer
* jouer
* manger
* dormir
* faire des bêtises

Dans un dictionnaire, comme le fait le langage Perl, nous pourrions ajouter des fonctions pour représenter ces actions. Mais ce ne serait pas très adapté :

````python
def manger(aliment):
    print(f"Miam, c'est bon le {aliment}")


personne = {"NOM": "Martin", "PRENOM": "Toto", "AGE": 10, "MANGER": manger}

personne["MANGER"]("chocolat")
````

Cette façon de faire est un peu tirée par les cheveux.

Les classes permettent de créer un nouveau type de données en regroupant dans une même structure:

* Le nom du type
* Ses *caractéristiques*, que l'on appelle aussi *attributs*, *propriétés*  
  Ce sont les valeurs qui le décrivent
* Les actions que l'on peut faire avec, que l'on appelle *méthodes*, *fonctions membres*

Pour créer une classe il convient d'utiliser le mot clef **class**, suivi de son nom.  Puis on démarre un bloc de code qui contiendra la description de la classe.

````python
class Vehicule:
    """On documente une classe comme une fonction
    Avec une docstring juste après son nom

    Une classe est une usine qui fabrique des variables
    """

    pass
````

Vous pouvez concevoir une classe comme une usine qui fabrique des variables.

Pour créer une variable du type de la classe, on appelle la classe comme une fonction :

````python
v1 = Vehicule()
print(type(v1))
print(type(v1) == Vehicule)
````

Mais que peut-on faire avec v1, il n'y a rien dans la classe ?

````python
print(dir(v1))
````

Pas grand chose effectivement. La fonction *dir* ne retourne que des noms que nous avons ignorés jusqu'ici.  Nous les présenterons dans quelques instants.



## Attributs de classe et d'instance

Commençons par ajouter des caractéristiques à nos véhicules.

Pour définir des caractéristiques partagées par toutes les variables de la classe, il convient de définir des variables dans la classe.

**Terminologie**: une variable d'une classe s'appelle aussi **instance** de la classe

````python
class Vehicule:
    identifiant = None
    couleur = "Blanc"
    poids = 1.0
````

On accède à ces attributs via la syntaxe `variable.nom_attribut`

````python
v1 = Vehicule()
print(dir(v1))
print("Couleur de v1:", v1.couleur)
````

Créons maintenant un second véhicule:

````python
v2 = Vehicule()
print("Couleur de v2:", v2.couleur)
````

Comme v2 est lui aussi blanc, notre apprenti stagiaire le confond tout le temps avec v1. Aussi pour qu'il cesse de les confondre, changeons la couleur de v1:

````python
v1.couleur = "Rouge"
print("Couleur de v1:", v1.couleur)
````

v2 est-il lui aussi devenu rouge ?

````python
print("Couleur de v2:", v2.couleur)
````

Non, il est resté blanc ! Tout va bien !

Imaginons maintenant qu'un vendeur de peinture arrive à l'usine et propose au directeur de l'usine qui fabrique les véhicules d'acheter sa nouvelle couleur *verte* bien plus jolie et bien moins onéreuse.

Ce que le directeur, habile en affaires, accepte:

````python
Vehicule.couleur = "Vert"
````

Ayant appris que de nouveaux véhicules verts étaient disponibles, nous nous empressons d'en acheter un troisième.

````python
v3 = Vehicule()
print("Couleur de v3:", v3.couleur)
````

v1 est-il devenu Vert lui aussi ?

````python
print("Couleur de v1:", v1.couleur)
````

Non, tout va bien !

Et v2 ?

````python
print("Couleur de v2:", v2.couleur)
````

Mais que s'est-il passé ?

* **Pourquoi** v1 est resté rouge ?
* **Pourquoi** v2 n'est pas resté blanc et est devenu vert ?

En fait, il y avait un piège dans notre énoncé: les attributs définis dans une classe sont des attributs qui appartiennent à la classe, ils décrivent la classe. On les appelle attributs statiques ou de classe.

Les variables fabriquées n'en possèdent pas leur propre copie !

Ce sont des attributs qui appartiennent à l'usine.

La couleur de la classe représente en quelque sorte le stock de peinture à l'usine.

Et les variables fabriquées ne possèdent finalement aucune information personnelle, ce sont des boîtes vides.

Quand nous demandons à Python la couleur de v1, il cherche dans v1 cette information qu'il ne trouve pas, aussi, avant de générer une erreur, il regarde dans la classe si cette information existe et il trouve la couleur du stock de peinture qu'il retourne.

![poo attr classe](images/poo_1.png)

Les attributs propres à une variable sont stockés dans le dictionnaire `__dict__` affiché par la fonction `dir`.  
Nous pouvons vérifier que `v2` ne possède aucune information.

````python
print(v2.__dict__)
````

Par contre lorsque nous écrivons `v1.couleur = 'Rouge'` nous créons dans la boîte `v1` une information couleur qui lui est propre. `v1` possède maintenant cet attribut qui lui est personnel.

![poo attribut instance](images/poo_2.png)

Nous pouvons vérifier que `v1` possède sa propre couleur:

````python
print("Attributs propres à v1:", v1.__dict__)
# alors que v2 n'en a pas:
print("Attributs propres à v2:", v2.__dict__)
````

Nous disposons donc de deux variables du même type qui n'ont pas les mêmes informations. Dans beaucoup de langages objets ceci n'est pas possible, comme Java ou C++.  D'ailleurs la première fois qu'un développeur venant de ces langages découvre cela en python **il crie à l'hérésie !**  

Mais dans la pratique c'est drôlement cool, et bien plus proche de la vie réelle:

* Vous pouvez acheter 2 voitures de la même marque et du même modèle et en personnaliser une avec une antenne CiBi, un aileron et des jantes aluminium, cela restera une voiture de la marque et du modèle, avec des choses en plus !

Peut-on ajouter à `v1` un attribut qui ne soit pas dans la classe `Vehicule` ?

````python
v1.nb_portes = 5
print("Nombre de portes de v1:", v1.nb_portes)
print("Attributs propres à v1:", v1.__dict__)
````

````python
print("Nombre de portes de v2:", v2.nb_portes)
````

Effectivement cela peut-être problématique, en général on évite de travailler ainsi, mais c'est bien pratique parfois comme vous le verrez avec Pandas.

Une variable peut avoir des attributs en plus, mais aussi les perdre !

````python
print("v1 a un accident, ses portes sont détruites...")
del v1.nb_portes

print("Attributs propres à v1:", v1.__dict__)
print("Nombre de portes de v1:", v1.nb_portes)
````

Et quand les problèmes arrivent, ils ne s'arrêtent pas...

La peinture de `v1` se dilue sous la pluie:

````python
del v1.couleur
````

Quelle est la couleur de v1 ? Erreur ?

````python
print("Couleur de v1:", v1.couleur)
````

Il restait celle à l'usine...

Mais l'entrepot prend feu, pas de chance...

````python
del Vehicule.couleur
````

````python
print("Couleur de v1:", v1.couleur)
````

## Attributs d'instance

Pour créer automatiquement des attributs à toutes les instances de la classe on le fait dans la méthode spéciale `__init__` qui doit être définie dans la classe.

Cette fonction est automatiquement appelée par Python dès qu'une variable est créée.  

On l'appelle *initialiseur*, c'est un peu comme le constructeur en Java/C++.

````python
class Vehicule:
    couleur = "Blanc"

    def __init__(self):
        """
        Quand on écrit
        v1 = Vehicule()

        Python crée v1 en mémoire puis exécute AUTOMATIQUEMENT

        Vehicule.__init__(v1)

        Le paramètre self représente donc la variable v1 manipulée/créée
        """
        print("Création du véhicule n°", id(self))
````

````python
v1 = Vehicule()  # exécute Vehicule.__init__(v1)
print("ID de v1:", id(v1))
````

Le nom et le rôle de la méthode `__init__` est imposé par Python.  

C'est le bon endroit pour initialiser les attributs des instances...

````python
class Vehicule:
    couleur = "Blanc"
    nb_vehicules_crees = 0

    def __init__(self):
        """
        Quand on écrit
        v1 = Vehicule()

        Python crée v1 en mémoire puis exécute AUTOMATIQUEMENT

        Vehicule.__init__(v1)

        Le paramètre self représente donc la variable v1 manipulée/créée
        """
        self.couleur = Vehicule.couleur
        self.identifiant = None
        self.poids = 1.0
        Vehicule.nb_vehicules_crees += 1
        self.numero_serie = Vehicule.nb_vehicules_crees
````

````python
v1 = Vehicule()
print(v1.__dict__)
````

````python
v2 = Vehicule()
print(v2.__dict__)
````

Maintenant, `v1` et `v2` possèdent leurs propres attributs dès leur création.

````python
v1.couleur = "Rouge"
Vehicule.couleur = "Vert"
v3 = Vehicule()

print("Couleur de v1:", v1.couleur)
print("Couleur de v1:", v2.couleur)  # ne devient pas Vert
print(
    "Couleur de v3:", v3.couleur
)  # mais c'est sa couleur personnelle, pas celle de la classe
````

Il est aussi possible de personnaliser la création des véhicules en passant des paramètres à la classe lors de leur création. Ces paramètres seront retransmis à la fonction `__init__`

````python
class Vehicule:
    couleur = "Blanc"
    nb_vehicules_crees = 0

    def __init__(self, couleur, identifiant):
        """
        Quand on écrit
        v1 = Vehicule(couleur='Orange', identifian='2048 BIT 75')

        Python crée v1 en mémoire puis exécute AUTOMATIQUEMENT

        Vehicule.__init__(v1, couleur='Orange', identifian='2048 BIT 75')
        """
        self.couleur = couleur
        self.identifiant = identifiant
        self.poids = 1.0
        Vehicule.nb_vehicules_crees += 1
        self.numero_serie = Vehicule.nb_vehicules_crees
````

````python
v1 = Vehicule(couleur="Noir", identifiant="K2000")

print(v1.__dict__)
````

Il est aussi possible d'ajouter des fonctions dans la classe.  

Elles doivent prendre un premier paramètre `self` qui représente l'instance manipulée.  Son nom est `self` par convention.




````python
class Vehicule:
    couleur = "Blanc"
    nb_vehicules_crees = 0

    def __init__(self, couleur, identifiant):
        """
        Quand on écrit
        v1 = Vehicule(couleur='Orange', identifian='2048 BIT 75')

        Python crée v1 en mémoire puis exécute AUTOMATIQUEMENT

        Vehicule.__init__(v1, couleur='Orange', identifian='2048 BIT 75')
        """
        self.couleur = couleur
        self.identifiant = identifiant
        self.poids = 1.0
        Vehicule.nb_vehicules_crees += 1
        self.numero_serie = Vehicule.nb_vehicules_crees

    def infos(self):
        """
        v1 = Vehicule()
        v1.infos() exécutera Vehicule.infos(v1)
        """
        print("Véhicule n°", self.numero_serie)
        print("  Couleur:", self.couleur)
        print("  Identifiant:", self.identifiant)
        print("  Poids:", self.poids)
````

````python
v1 = Vehicule(couleur="Jaune", identifiant="La coccinelle")
v1.infos()
````

Vous pouvez aussi écrire `Vehicule.infos(v1)` mais la notation est plus laborieuse et non recommandée : elle inclut le nom de la classe, si celui-ci change il faudra modifier tous les appels de méthodes.

## Méthodes spéciales

Tous les autres noms que vous observez avec la fonction `dir` représentent des noms de fonctions ou attributs spéciaux qui implémentent des comportements particulier dans la mécanique objets de votre classe.

Ils sont définis dans la documentation Python au chapitre [Data Model](https://docs.python.org/3/reference/datamodel.html) que nous vous invitons à lire au moins une fois.

Voici quelques exemples de fonctions spéciales:

* « `__str__` » pour convertir votre objet en chaîne  
  Exemple : print("mon objet :%s" % mon_objet)
* « `__bool__` » en Python 3 et « `__nonzero__` » en Python 2  
  permet de savoir si votre objet est évalué comme valant True ou False dans une expression booléenne
* « `__del__` » : Le destructeur. C'est le pendant du constructeur.  
  Cette méthode est appelée quand votre objet est détruit.  
  Exemple : del v1 ou v1 = None
* « `__add__` », « `__sub__` », « `__mul__` » : Pour utiliser les opérateurs mathématiques '+', '-' et '*' entre 2 objets
