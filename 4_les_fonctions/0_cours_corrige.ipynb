### La suite de Fibonacci
La suite de Fibonacci est une suite de nombres dans laquelle chaque nombre est la somme des deux précédents. La suite commence par 0 et 1. Voici les premiers termes de la suite :
0, 1, 1, 2, 3, 5, 8, 13 ...

Elle s'écrit comme suit: `u(n) = u(n-1) + u(n-2)`
et `u(0) = 0`, `u(1) = 1`

1. Ecrivez une fonction `fibonacci(n)` qui prend un entier `n` en argument et renvoie le n-ième terme de la suite de Fibonacci. La fonction doit etre documentée avec une docstring et doit utiliser des annotations de type.
2. Ajouter une fonction `export_fibonacci_result(n: int, filename="resultat.txt") -> None` qui prend un entier `n` en argument et qui appelle la fonction `fibonacci(n)` pour écrire le résultat dans un fichier texte `resultat.txt`: `Le résultat de fibonacci(n) est : <resultat>`.
3. Ajouter une fonction `def display_file_content(filename="resultat.txt") -> None` qui lit le fichier `resultat.txt` et affiche le contenu à l'écran.
4. Ajouter une fonction `main(n: int) -> None` qui prend un entier `n` en argument et affiche le contenu de `resultat.txt`à l'écran !

##### Question ouverte:
- Est-ce que ces trois fonctions respectent le principe de responsabilité unique ?

````python
# Exercice 1 - La suite de Fibonacci
def fibonacci(n: int) -> int:
    """Calcul la suite de Fibonacci
    u(n+2) = u(n+1) + u(n)
    Sachant que u(0) = 0 et u(1) = 1

    :param n: L'entier représentant le n-ième terme de la suite de Fibonacci
    :reutrn: La valeur de la suite de Fibonacci
    """
    if n == 0:
        return 0

    if n == 1:
        return 1

    fibonacci_sequence = [0, 1]

    if n >= 2:
        for i in range(2, n + 1):
            fibonacci_sequence.append(
                fibonacci_sequence[i - 2] + fibonacci_sequence[i - 1]
            )

        return fibonacci_sequence[-1]
````

````python
resultat = fibonacci(n=13)
````

````python
def export_fibonacci_result(n: int, filename="resultat.txt") -> None:
    """Ecrire le n-ième terme de la suite de Fibonacci et écrit le résultat dans un fichier
    texte: resultat.txt: Le résultat de fibonacci(n) est : <resultat>
    """
    resultat = fibonacci(n=n)
    with open(filename, "w") as f:
        f.write(f"Le résultat de fibonacci({n}) est : {resultat}")
````

````python
export_fibonacci_result(n=13)
````

````python
def display_file_content(filename="resultat.txt") -> None:
    """Lire le fichier filename et print le contenu"""
    with open(filename, "r") as f:
        content = f.readlines()[0]

    print(content)
````

````python
def main(n: int) -> None:
    """La fonction main calcule le n-ième terme de la suite de Fibonacci et écrit le résultat dans un fichier
    et affiche le contenu du fichier à l'écran

    :param n: L'entier représentant le n-ième terme de la suite de Fibonacci"""
    export_fibonacci_result(n=n)
    display_file_content()


main(13)
````

### Manipulation de Chaînes de Caractères avec Gestion des Exceptions
Vous allez créer une fonction qui traite un email reçu et gère les exceptions potentielles, comme des formats incorrects ou des valeurs manquantes.

1. Écrivez une fonction traiter_email(email: str) -> dict qui prend une chaîne de caractères représentant un email en argument et renvoie un dictionnaire contenant les informations extraites.

Le format d'un email est le suivant :
```
"Sujet": <sujet>\n
"De": <expéditeur>\n
"Contenu": <contenu>
```

Le format de dictionnaire de sortie est:
```
{
    "sujet": <sujet>,
    "expediteur": <expéditeur>,
    "contenu": <contenu>
}
```

Tips:
* Pensez à split('\n') pour créer une liste de lignes !
* Pensez à split(':') pour séparer les clés et les valeurs !
* Pensez à isinstance() pour vérifier le type de l'argument !

La fonction doit gérer les exceptions suivantes :

* ValueError : Si l'email ne contient pas les informations attendues
    * Si l'email ne contient pas assez de lignes, information[1].split(":")[1].strip() levera un `IndexError`
* TypeError : Si l'argument n'est pas une chaîne de caractères.

La fonction doit être documentée avec une docstring et utiliser des annotations de type.

2. Ajoutez une fonction valider_dictionnaire(donnees: dict) -> bool qui prend un dictionnaire en argument et vérifie s'il contient certaines clés obligatoires (par exemple, "sujet", "expediteur", et "contenu"). La fonction doit gérer les exceptions suivantes :

* KeyError : Si une clé obligatoire est manquante.
* TypeError : Si l'argument n'est pas un dictionnaire.

3. Ajoutez une fonction main() qui appelle les fonctions traiter_email et valider_dictionnaire avec des exemples de données. La fonction main() doit être appelée si le script est exécuté directement.

````python
email = """Sujet: Réunion de projet MIMO
De: louis@giron-dom.eu
Contenu: Bonjour à tous, la réunion de projet MIMO est prévue pour demain à 10h!
"""
````

````python
# 1.
def traiter_email(email: str) -> dict:
    """Prend une chaîne de caractères représentant un email en argument et renvoie un dictionnaire
    contenant les informations extraites.
    :param email: str l'email à traiter
    :return: les informations extraites du texte
    """
    if not isinstance(email, str):
        raise TypeError("email n'est pas un string valide")

    information = email.split("\n")
    try:
        sujet = information[0].split(":")[1].strip()
        expediteur = information[1].split(":")[1].strip()
        contenu = information[2].split(":")[1].strip()
    except IndexError as e:
        raise ValueError("email ne contient pas les informations attendues")

    res = {
        "sujet": sujet,
        "expediteur": expediteur,
        "contenu": contenu,
    }
    if sujet is None or expediteur is None or contenu is None:
        raise ValueError("email ne contient pas les informations attendues")

    return res
````

````python
traiter_email(email=email)
````

````python
traiter_email(email=2)
````

````python
traiter_email(email="not a valid email")
````

````python
# 2.
def valider_dictionnaire(donnees: dict) -> bool:
    """vérifie s'il contient certaines clés obligatoires ("sujet", "expediteur", et "contenu")
    :param donnees: Le dictionnaire à vérifier
    :return: Booléen : le dictionnaire est valide ou non
    """
    if not isinstance(donnees, dict):
        raise ValueError("donnees n'est pas un dictionnaire valide")

    keys = ["sujet", "expediteur", "contenu"]
    for key in keys:
        if key not in donnees.keys():
            raise KeyError("donnees ne contient pas les valeurs attendues")
    return True
````

````python
valider_dictionnaire(traiter_email(email=email))
````

````python
valider_dictionnaire({"test": "error"})
````

## Exercice 🐍 - La suite de Syracuse

La suite de Syracuse est une suite de nombres définie de la manière suivante :
- Si n est pair, le terme suivant est n/2
- Si n est impair, le terme suivant est 3n + 1
- La suite se termine lorsque le terme atteint 1.

1. Ecrivez une fonction `syracuse_sequence(n)` qui prend un entier `n` en argument et renvoie la suite de Syracuse sous forme de liste. La fonction doit être documentée avec une docstring et doit utiliser des annotations de type.
2. Ecrivez une fonction `syracuse_convergence_index(n)` qui prend un entier `n` en argument et renvoie l'index du premier terme de la suite de Syracuse qui atteint 1. La fonction doit être documentée avec une docstring et doit utiliser des annotations de type.

````python
def syracuse_sequence(n: int) -> tuple[list[int], int]:
    """
    La suite de Syracuse est une suite de nombres définie de la manière suivante :

    Si n est pair, le terme suivant est n/2
    Si n est impair, le terme suivant est 3n + 1
    La suite se termine lorsque le terme atteint 1.

    :param n: L'entier représentant le premier terme de la suite de Syracuse
    :return: tuple (La séquence de la suite de syracuse à partir de n, le nombre d'itération avant d'atteindre 1)
    """
    iterations = 0
    s_n = [n]
    s = 0
    while s != 1:
        if s_n[iterations] % 2 == 0:
            s = int(s_n[iterations] / 2)
        elif s_n[iterations] % 2 == 1:
            s = int(3 * s_n[iterations] + 1)

        s_n.append(s)
        iterations += 1

    return s_n
````

````python
syracuse_sequence(15)
````

````python
def syracuse_convergence_index(n: int) -> int:
    """
    La fonction syracuse_convergence_index prend un entier n en entrée et renvoie l'index du premier
    terme de la suite de Syracuse qui est égal à 1.

    :param n: L'entier représentant le premier terme de la suite de Syracuse
    :return: L'index du premier terme de la suite de Syracuse qui est égal à 1
    """
    s_n = syracuse_sequence(n=n)
    return s_n.index(1)
````

### La factorielle d'un nombre

La factorielle d'un nombre entier positif n est le produit de tous les entiers positifs inférieurs ou égaux à n. Elle est notée n! et se définit comme suit :
- n! = n × (n - 1)! pour n > 0
- 0! = 1

donc 5! = 1 * 2 * 3 * 4 * 5

1. Écrivez une fonction `factorielle(n: int) -> int` qui prend un entier positif `n` en argument et renvoie la factorielle de `n`. La fonction doit être documentée avec une docstring et utiliser des annotations de type.
2. Écrivez une fonction `factorielle_recursive(n: int) -> int` qui utilise la récursivité pour calculer la factorielle de `n`. La fonction doit être documentée avec une docstring et utiliser des annotations de type.
3. Écrivez un test unitaire qui vérifie que la fonction `factorielle_recursive(n)` renvoie le bon résultat pour plusieurs valeurs de `n`. Vous testerez pour cela:
* Factoriel de 0 == 1
* Factoriel de 1 == 1
* Factoriel de 3 == 6
* Factoriel de 5 == 120

````python
# 1.
def factorielle(n: int) -> int:
    """Renvoie la factorielle de n
    :param n: entier positif sur lequel calculer la factorielle
    :return: la valeur de la factorielle de n
    """
    if n < 0:
        raise ValueError("N doit être un entier positif ou nul, dans N")

    if n == 0:
        return 1

    value = 1
    for i in range(1, n + 1):
        value *= i
    return value
````

````python
for i in range(5):
    print(f"Factoriel de {i} == {factorielle(n=i)}")
````

````python
# 2.
def factorielle_recursive(n: int) -> int:
    """Renvoie la factorielle de n avec une approche recursive
    :param n: entier positif sur lequel calculer la factorielle
    :return: la valeur de la factorielle de n
    """
    if n < 0:
        raise ValueError("N doit être un entier positif ou nul, dans N")
    if n == 0:
        return 1  # Condition d'arrêt

    return n * factorielle_recursive(n - 1)
````

````python
for i in range(6):
    print(f"Factoriel de {i} == {factorielle_recursive(n=i)}")
````

````python
# 3. Test unitaire
def test_factorielle_recursive():
    # Test de la factorielle de 0
    assert factorielle_recursive(0) == 1, "Erreur: la factorielle de 0 est 1"

    # Test de la factorielle de 1
    assert factorielle_recursive(1) == 1, "Erreur: la factorielle de 1 est 1"

    # Test de la factorielle de 3
    assert factorielle_recursive(3) == 1 * 2 * 3, "Erreur: la factorielle de 3 est 6"

    # Test de la factorielle de 5
    assert factorielle_recursive(5) == 1 * 2 * 3 * 4 * 5, (
        "Erreur: la factorielle de 5 est 120"
    )


test_factorielle_recursive()
````

## Exercice 🐍 – Analyse de données brutes

Vous travaillez sur une API météorologique (des températures) pour prédire la production éolienne de la semaine.

Avant de les exploiter, vous devez :

1. Filtrer les mesures valides : éliminer toutes les valeurs supérieures à 50°C et inférieures à -30°C, elles sont considérées comme des erreurs de mesure.

2. Appliquer une transformation linéaire : multiplier chaque mesure valide par 0.9 (facteur d’ajustement du capteur)

3. Calculer un indicateur global : la moyenne des mesures valides

Bonus – Version paresseuse : réécrivez l’étape 2 avec un générateur qui transforme les valeurs à la volée, sans créer de nouvelle liste en mémoire.


````python
donnees_temp = [-5, -3, 1, 5, 7, -34, -52, 15, 22, 25, 30, 28, 25, 89, 67, 23, 16]

# A vous de jouer!
````

````python
# 1. Filtrer
donnees_valides = list(filter(lambda x: -30 <= x <= 50, donnees_temp))
donnees_valides
````

````python
# 2. Ajuster
donnees_ajustees = list(map(lambda x: x * 0.9, donnees_valides))
donnees_ajustees
````

````python
# 3. Indicateur
from functools import reduce

indicateur = reduce(lambda x, y: x + y, donnees_ajustees) / len(donnees_ajustees)
indicateur
````

````python
# 4. Bonus – Version paresseuse : réécrivez l’étape 1 et 2 avec un générateur qui
# transforme les valeurs à la volée, sans créer de nouvelle liste en mémoire.


def generate_donnees_ajustees(donnees):
    for d in donnees:
        if -30 <= d <= 50:
            yield d * 0.9
````

````python
donnees_ajustees_gen = generate_donnees_ajustees(donnees_temp)

for d in donnees_ajustees_gen:
    print(d)
````

````python
indicateur = sum([x for x in generate_donnees_ajustees(donnees_temp)]) / len(
    [x for x in generate_donnees_ajustees(donnees_temp)]
)
indicateur
````
