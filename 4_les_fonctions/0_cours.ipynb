Une fonction est un bloc de code avec un but spécifique auquel vous pouvez donner un nom. Quand vous appelez cette fonction, vous exécutez le code qu’elle contient. Les fonctions vous laissent saisir des paramètres pour exécuter le même code sur différentes valeurs.  

Il y a différents types de fonctions dans Python :

Les fonctions intégrées fournies avec Python.
Les fonctions définies par l’utilisateur que les développeurs (vous !) créent. 
Vous avez déjà utilisé une fonction intégrée ! range() est une fonction dans laquelle vous pouvez saisir un nombre et elle renverra ou répondra avec une séquence allant de 0 à -1.

Pensez à une fonction comme à un moyen de réutiliser un ensemble d’instructions répétables. Vous la définissez avec le mot-clé def, le nom de la fonction, des parenthèses et deux-points:. Si la fonction a besoin d’un paramètre ou plus, il faut les saisir à l’intérieur des parenthèses, séparés par des virgules. Admettons que vous voulez ajouter deux nombres ensemble. L’extrait de code ci-dessous est une méthode ``add()`` qui prend deux nombres comme paramètres et renvoie la somme. 

```python
def add(x, y):
    return x + y
```
Vous pouvez appeler la fonction en utilisant son nom et en lui passant les valeurs que vous voulez additionner. Par exemple, si vous voulez additionner 2 et 3, vous pouvez faire ceci : 

```python
result = add(2, 3)
print(result)
```
Cela affichera 5. Vous pouvez également appeler la fonction avec des variables, comme ceci : 

```python
a = 2
b = 3
result = add(a, b)
print(result)
```

### Quand utiliser les fonctions et pourquoi ?

Quand on écrit beaucoup de code, on s’y perd et on fait des erreurs facilement entre les différentes fonctionnalités en cours. Les fonctions vous aident à séparer le code en sections plus petites. Comme ça vous gardez le fil sur ce que chaque partie est censée faire.

Vous en sortirez avec un code mieux écrit, mieux structuré et plus lisible.

Primordial de bien structurer son code, pour soi et aussi pour les personnes qui devront le reprendre après vous.

Plusieurs principes de programmation existent pour vous aider à écrire du code propre et lisible:

- Respectez les standards du code: Python a des conventions de nommage et de style (PEP 8) que vous devriez suivre pour rendre votre code cohérent et lisible.
    - Identation: 4 espaces par niveau d'indentation
    - Espaces autour des opérateurs: 1 espace avant et après les opérateurs binaires
    - Longueur des lignes: 79 caractères maximum
    - Nom de variables: Utilisez des noms significatifs et en minuscules, avec des underscores: snake_case pour séparer les mots (ex: `ma_variable`).

- Le principe du **"DRY" (Don't Repeat Yourself)** : Évitez de répéter le même code plusieurs fois. Utilisez des fonctions pour encapsuler le code réutilisable.

- Le principe de **responsabilité unique** : Chaque fonction doit avoir une seule responsabilité ou tâche. Cela rend le code plus facile à comprendre et à maintenir.

- **La documentation: commenter le code** : Utilisez des commentaires pour expliquer le but de chaque fonction et son fonctionnement. Cela aide les autres (et vous-même) à comprendre le code plus tard.
   - Maximum 72 caractères pour les commentaires & docstrings
   - Plusieurs conventions existent: https://stackoverflow.com/questions/3898572/what-are-the-most-common-python-docstring-formats
   - On verra dans le cours la convention ``reStructuredText (reST)``

- Annotations de type (Typage des variables): Depuis Python 3.5, vous pouvez annoter vos fonctions avec des types pour indiquer quel type de données elles attendent et renvoient. Cela peut aider à la lisibilité et à la détection des erreurs.

## Créer une fonction

Pour créer une fonction:
* On utilise le mot clef **def** qui signifie *define*
* Suivi du nom de la fonction
* De parenthèses pour les éventuels paramètres 
* Puis de la marque de début de bloc `:`  
* Tout ce qui est indenté après cette ligne de définition fait partie du code de la fonction
* La fonction peut renvoyer une valeur avec le mot clé `return` suivi de la valeur à renvoyer
* Si la fonction ne renvoie rien, elle renvoie `None` par défaut

Pour appeler la fonction on écrit son nom suivi de parenthèses `()`

````python
def add_integers(a, b):
    # Identation: 4 espaces par niveau d'indentation
    # Longueur des lignes: 79 caractères maximum
    result = (
        a + b
    )  # Espaces autour des opérateurs: 1 espace avant et après les opérateurs binaire
    # Nom de variables: Utilisez des noms significatifs et en minuscules
    print("Addition de", a, "et", b, "est", result)
    return result


# Exemple d'utilisation de la fonction
add_integers(3, 5)
add_integers(10, 20)
add_integers(0, 0)
add_integers(-5, 5)

# On respecte ici les standards du code: Python a des conventions de nommage et de
# style (PEP 8):
# - Identation: 4 espaces par niveau d'indentation
# - Espaces autour des opérateurs: 1 espace avant et après les opérateurs binaires:
# - Longueur des lignes: 79 caractères maximum
# - Nom de variables: Utilisez des noms significatifs et en minuscules, avec des underscores: snake_case pour séparer les mots (ex: `ma_variable`).
````

### Le principe du DRY

Imaginons que vous devez calculer la moyenne de plusieurs listes de nombres. Au lieu de répéter le code de calcul de la moyenne pour chaque liste, vous pouvez créer une fonction réutilisable

````python
# Sans DRY
liste1 = [1, 2, 3, 4, 5]
liste2 = [10, 20, 30, 40, 50]

# Calcul de la moyenne pour liste1
somme1 = sum(liste1)
moyenne1 = somme1 / len(liste1)

# Calcul de la moyenne pour liste2
somme2 = sum(liste2)
moyenne2 = somme2 / len(liste2)

print("Moyenne de liste1 :", moyenne1)
print("Moyenne de liste2 :", moyenne2)
````

````python
# Avec DRY
def calculer_moyenne(liste):
    return sum(liste) / len(liste)


liste1 = [1, 2, 3, 4, 5]
liste2 = [10, 20, 30, 40, 50]

moyenne1 = calculer_moyenne(liste1)
moyenne2 = calculer_moyenne(liste2)

print("Moyenne de liste1 :", moyenne1)
print("Moyenne de liste2 :", moyenne2)
````

## Documentation de la fonction

On peut aussi ajouter une **docstring** (documentation de la fonction) entre triple guillemets pour expliquer ce que fait la fonction. C'est une bonne pratique pour rendre le code plus lisible et compréhensible.

````python
def is_number_prime(n: int) -> bool:
    """Vérifie si un nombre est premier.
    Un nombre est premier s'il est supérieur à 1 et n'a pas de diviseur
    :param n: Le nombre à vérifier
    :return: True si le nombre est premier, False sinon
    """
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
````

On remarquera ici plusieurs choses:
* La docstring: documente la fonction
* Le typage des arguments: `n: int` signifie que `n` doit être un entier
* Le typage du retour: `-> int` signifie que la fonction renvoie un entier

Mais surtout:
* Les multiples return !

### Comment ca fonctionne ?

La fonction `is_number_prime` vérifie si un nombre est premier avec l'approche conditionnelle vu au chapitre précédent.
Si une condition est vérifiée, le bloc de code comprenant le `return` est exécuté et la fonction renvoie la valeur associée sans exécuter le reste du code!

Donc les multiples `return` permettent de sortir de la fonction à différents moments selon les conditions rencontrées.

````python
print("Est-ce que 5 est un nombre premier?", is_number_prime(5))
print("Est-ce que 4 est un nombre premier?", is_number_prime(4))
print("Est-ce que 169 est un nombre premier?", is_number_prime(169))
````

Python stocke cette chaîne dans une variable spéciale: l'attribut `__doc__` de la fonction.

Elle est utilisée pour générer l'aide en ligne

### Paramètres positionnels ou nommés

Il est possible de passer les paramètres d'une fonction sans respecter l'ordre dans lequel ils ont été définis.

````python
def operation(a: int, b: int, op: str) -> int:
    """
    Effectue une opération arithmétique sur deux entiers.
    :param a: Le premier entier.
    :param b: Le deuxième entier.
    :param op: L'opération à effectuer (sous forme de chaîne de caractères).
    :return: Le résultat de l'opération."""
    print(f"a={a}, b={b}, op={op}")
    match op:
        case "+":
            return a + b
        case "-":
            return a - b
        case "*":
            return a * b
        case "/":
            return a / b
        case _:
            print("Opération non valide")
````

````python
# Positionnels
operation(10, 5, "*")  # a = 10, b = 5, op = '*'
````

````python
# Nommés
operation(a=10, b=5, op="-")  # a = 10, b = 5, op = '-'

# On peut, en nommant les arguments, changer l'ordre
operation(op="/", b=5, a=10)  # a = 10, b = 5, op = '/'
````

Ok, mais que se passe-t-il dans ce cas ?

````python
operation(6, a=5, op="*")
````

Et dans celui-ci ?

````python
operation(a=5, 6, "*")
````

**Explications**

Dans le cas `operation(6, a=5, op='*')` l'erreur se comprend assez bien: la valeur 6, paramètre positionnel, est affectée à `a` mais l'on précise aussi que `a` reçoit 5. Aussi *Python* est perdu et refuse de continuer.

Dans le second cas, l'erreur peut sembler moins évidente, mais elle repose sur une règle fondamentale en Python :

**Les paramètres positionnels doivent toujours précéder les paramètres nommés.**

### Pourquoi cette règle ?
Lorsque vous mélangez des paramètres positionnels et nommés, Python doit être capable d'associer chaque valeur à son paramètre correspondant. Si un paramètre nommé apparaît avant un paramètre positionnel, l'ordre devient ambigu et Python ne peut plus déterminer correctement à quel paramètre chaque valeur correspond.

## La gestion des erreurs

Il est primordial de gérer les erreurs dans votre code. Vous pouvez utiliser des blocs `try/except` pour capturer les exceptions et éviter que le programme ne plante!

Il est possible également de lever des exceptions personnalisées avec le mot clé `raise`. Cela peut être utile pour signaler des erreurs spécifiques dans votre code. Par exemple, si vous avez une fonction qui attend un entier positif, vous pouvez lever une exception si l'utilisateur entre un nombre négatif.

````python
1 / 0
````

````python
def divide(a: float, b: float) -> float:
    """Divise a par b.
    :param a: Le numérateur
    :param b: Le dénominateur
    :return: Le résultat de la division
    :raises ZeroDivisionError: Si b est égal à 0
    """
    if b == 0:
        raise ZeroDivisionError("Attention: le dénominateur ne peut pas être zéro.")
    return a / b
````

````python
divide(10, 2)  # 5.0
````

````python
divide(10, 0)  # Lève une exception ZeroDivisionError
````

````python
# Il est possible de capturer une exception "Générale" avec Exception, mais ce n'est pas une bonne pratique
# car cela peut masquer d'autres erreurs.
# Il est préférable de capturer des exceptions spécifiques.
````

````python
def divide(a: int, b: int) -> float:
    """Divise a par b.
    :param a: Le numérateur
    :param b: Le dénominateur
    :return: Le résultat de la division
    """
    try:
        return a / b
    except Exception:
        print("Erreur: Il n'est pas possible de diviser par 0.")
        return None
````

````python
# Il est possible de lever plusieurs exceptions & d'utiliser la syntaxe "as" pour capturer l'exception et l'afficher
def divide(a: int, b: int) -> float:
    """Divise a par b.
    :param a: Le numérateur
    :param b: Le dénominateur
    :return: Le résultat de la division
    """
    try:
        return a / b
    except ZeroDivisionError as e:
        print(f"Erreur: {e}, l'opération a échoué, division par zéro")
        return None
    except TypeError as e:
        print(f"Erreur: {e}, l'opération a échoué, mauvais type")
        return None
    except Exception as e:
        print(e)
````

## Tests unitaires

````python
def inverser_chaine(chaine: str) -> str:
    """Inverse la chaîne de caractères donnée.
    :param chaine: La chaîne à inverser
    :return: La chaîne inversée
    """
    return chaine[::-1]
````

````python
def test_inverser_chaine():
    """Test de la fonction inverser_chaine."""
    assert inverser_chaine("Python") == "nohtyP"
    assert inverser_chaine("") == ""
    assert inverser_chaine("12345") == "54321"
    print("Tous les tests ont réussi.")


test_inverser_chaine()
````

````python
def wrong_inverser_chaine(chaine: str) -> str:
    """Inverse la chaîne de caractères donnée.
    :param chaine: La chaîne à inverser
    :return: La chaîne inversée
    """
    # Mauvaise implementation
    return chaine
````

````python
def test_wrong_inverser_chaine():
    """Test de la fonction inverser_chaine."""
    expected = "nohtyP"
    actual = wrong_inverser_chaine("Python")
    error_msg = f"Test échoué pour 'Python', attendu: {expected}, got {actual}"
    assert actual == expected, error_msg
    print("Tous les tests ont réussi.")


test_wrong_inverser_chaine()
````

````python
# Ecrire un test unitaire est primordial pour s'assurer que le code fonctionne comme prévu.
# Développement TDD: Test driven development


def suite_de_conway(n: int) -> list:
    """La suite de Conway est une suite d'entiers étudiée en 1986 par le mathématicien britannique John Horton Conway,
    initialement sous le nom de « suite audioactive ». Elle est également connue sous le nom anglais de Look and Say
    sequence (« suite regarde et dis »).
    Dans cette suite autoréférente, chaque terme se détermine en annonçant les chiffres formant le terme précédent.

    :param n: Le nombre de termes à générer
    :return: La suite de Conway sous forme de liste

    Exemple:
    X_0 = 1
    X_1 = 11 (X_0 est égal à 1 "1")
    X_2 = 21 (X_1 est égal à 2 "1")
    X_3 = 1211 (X_2 est égal à 1 "2" et 1 "1")
    X_4 = 111221 (X_3 est égal à 1 "1", 1 "2" et 2 "1")
    ...
    """
    x_0 = 1
    seq = [x_0]

    for i in range(1, n + 1):
        # On convertit le nombre en chaîne de caractères pour le traiter
        previous_term = str(seq[i - 1])
        next_term = ""

        # On compte les occurrences de chaque chiffre consécutif: initialisation à 1
        count = 1
        for j in range(1, len(previous_term)):
            # On compare le chiffre actuel avec le précédent
            # Si ils sont égaux, on incrémente le compteur
            if previous_term[j] == previous_term[j - 1]:
                count += 1
            else:
                # Sinon, on ajoute le compteur et le chiffre précédent à la chaîne
                next_term += f"{count}{previous_term[j - 1]}"
                count = 1

        # On ajoute le dernier groupe de chiffres (e.g. "11" ou "22")
        next_term += f"{count}{previous_term[-1]}"
        seq.append(int(next_term))
    return seq
````

## Fonctions récursives

Une fonction récursive est une fonction qui s'appelle elle-même pour résoudre un problème. Cela peut être utile pour des problèmes qui peuvent être divisés en sous-problèmes similaires.

Il y a un point important dans l'écriture de ces fonctions, il faut une condition pour qu'elles cessent de s'appeler à un moment donné, sinon elles boucleront à l'infini

Exemples:
- Compte à rebours
- Fibonacci ...


````python
import time


def compte_a_rebours(n: int) -> None:
    """Compteur à rebours de n à 0
    :param n: Le nombre de départ
    :return: None
    """
    for i in range(n, 0, -1):
        print(i)
        time.sleep(0.5)  # Pause d'une seconde entre chaque nombre
    print("Partez!")


compte_a_rebours(5)
````

Mais nous pourrions aussi jouer sur le fait qu'elle affiche un nombre puis le suivant.

Pourquoi ne pas lui demander de se rappeler elle-même avec le nombre suivant pour l'afficher à son tour ?

````python
def compte_a_rebours(n: int) -> None:
    """Compteur à rebours de n à 0
    :param n: Le nombre de départ
    :return: None
    """
    print(n)
    # time.sleep(0.5)  # Pause d'une seconde entre chaque nombre
    compte_a_rebours(n - 1)


compte_a_rebours(5)
````

Lorsque vous exécutez le code ci-dessus, nous obtenons une erreur **Recursion error**

Le problème est que la fonction se rappelle à l'infini.

Mais la mémoire de l'ordinateur est limitée en taille et il ne peut pas gérer une infinité d'appels.

Il convient de lui préciser quand s'arrêter...

````python
def compte_a_rebours_recursive(n: int) -> None:
    """Compteur à rebours de n à 0
    :param n: Le nombre de départ
    :return: None
    """
    if n > 0:
        print(n)
        time.sleep(0.5)  # Pause d'une seconde entre chaque nombre
        compte_a_rebours_recursive(n - 1)
    else:
        print("Partez!")  # Condition d'arrêt de la récursion (n == 0)


compte_a_rebours_recursive(5)
````

````python
# Autre exemple, la suite de Fibonacci
# On a vu dans l'exercice précédent que la suite de Fibonacci est définie par
# la relation de récurrence suivante:
# u(n) = u(n-1) + u(n-2)
# avec u(0) = 0 et u(1) = 1


def fibonacci_recursive(n: int) -> int:
    """Calcul la somme de la suite de Fibonacci de manière récursive.
    :param n: Le rang de la suite de Fibonacci
    :return: La valeur de la suite de Fibonacci à ce rang
    """
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)


# Exemple d'utilisation de la fonction
for i in range(10):
    print(f"Fibonacci({i}) = {fibonacci_recursive(i)}")
````

#### Les fonctions anonymes (lambda)

````python
# Fonction lambda pour calculer le carré d'un nombre
carre = lambda x: x**2

carre(5)  # 25
````

````python
# On utilise les fonctinos lambda pour des opérateurs simples
# en général avec map, filter ou reduce

# map() applique une fonction à chaque élément d'une liste
nombres = [1, 2, 3, 4, 5]

# On utilise lambda pour élever chaque nombre au carré
carres = list(map(lambda x: x**2, nombres))

print(carres)  # Affiche [1, 4, 9, 16, 25]
````

````python
# filter() permet de filtrer les éléments d'une liste selon un critère
nombres = [1, 2, 3, 4, 5, 6]

# On garde seulement les nombres pairs
pairs = list(filter(lambda x: x % 2 == 0, nombres))

print(pairs)  # Affiche [2, 4, 6]
````

````python
from functools import reduce

# reduce() combine les éléments d'une liste avec une fonction binaire
nombres = [1, 2, 3, 4, 5]

# On calcule le produit de tous les éléments
produit = reduce(lambda x, y: x * y, nombres)

print(produit)  # Affiche 120 (1*2*3*4*5)

# reduce() applique la fonction de gauche à droite : ici, on multiplie chaque
# nombre successivement.
````

````python
letters = ["m", "a", "s", "t", "e", "r", "-", "m", "i", "m", "o"]

# On utilise lambda pour concaténer les lettres
mot = reduce(lambda x, y: x + y, letters)

print(mot)  # Affiche "master-mimo"
````

#### Les générateurs

Les générateurs sont des fonctions qui permettent de créer des itérateurs qui permettent de drastiquement réduire la mémoire utilisée par le programme. Ils sont créés en utilisant le mot-clé `yield` au lieu de `return`. Lorsqu'une fonction utilise `yield`, elle renvoie un générateur qui peut être itéré pour obtenir les valeurs une par une, **au lieu de les stocker toutes en mémoire**.

````python
def my_generator():
    """Un générateur simple qui produit les nombres de 0 à 4"""
    yield 0
    yield 1
    yield 2
    yield 3
    yield 4


# Utilisation du générateur
for i in my_generator():
    print(i)  # Affiche 0, 1, 2, 3, 4
````

````python
size = 0

# Combien de mémoire est utilisée pour stocker une liste de 1 million d'entiers?

import sys

for i in range(1_000_000):
    size += sys.getsizeof(i)

print(
    f"Nombre d'octets utilisés pour stocker 1 million d'entiers: {size} octets et donc {size / 1_000_000} Mo"
)

# 28000000 octets, c'est énorme !
````

````python
# Au lieu de stocker les entiers en mémoire, on peut les générer à la volée avec
# un générateur


def generate_numbers(n: int):
    """Générateur d'entiers de 0 à n-1"""
    for i in range(n + 1):
        yield i


generator = generate_numbers(1_000_000)

sys.getsizeof(generator)  # Affiche 112 octets

# 200 octets, c'est beaucoup moins que 28000000 octets!
````

````python
for g in generator:
    if g > 999_997:
        print(g)
````

````python
# Une fois appelée, le générateur ne peut plus être utilisé
# On peut le réinitialiser en le redéfinissant mais pas le réutiliser
for g in generator:
    print(g)
````

Pourquoi le générateur est épuisé ! Il n'y a plus de valeur à renvoyer. Vous pouvez le réinitialiser en le redéfinissant ou en le recréant.
